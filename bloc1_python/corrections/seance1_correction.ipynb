### Exercice 2 — Bénéfice et retour sur investissement

> **Votre mission :**
> - Une campagne coûte 3 200 € et rapporte 8 000 €.
> - Calculer le bénéfice dans `benefice`, puis le ROI dans `roi` — le bénéfice rapporté au coût, donc `benefice / cout`.
> - Arrondir le ROI à 2 décimales dans `roi_arrondi`.

```python
cout = 3200
revenu = 8000

benefice = revenu - cout   ## dans cet ordre : ce qui rentre moins ce qui sort
roi = benefice / cout      ## un ratio, donc sans unite

# round(valeur, 2) arrondit a deux chiffres apres la virgule
roi_arrondi = round(roi, 2)

print(benefice, "euros de benefice, ROI de", roi_arrondi)
```

### Question 7 — Le rapport complet d'une campagne

> **Votre mission :**
> - Une campagne email : 20 000 envois, 500 € de coût, 22 % d'ouvertures, 3 % de clics parmi les ouvertures, 18 € de panier moyen par clic.
> - Calculer le nombre d'ouvertures, le nombre de clics, le revenu, le bénéfice et le ROI.
> - Afficher un rapport de cinq lignes avec des `f-strings`.
> - *Nouveau :* dans un f-string, `{valeur:.2f}` arrondit à deux décimales, `{valeur:.0f}` en supprime, `{valeur:.0%}` met en pourcentage.

```python
envois = 20000
cout = 500
taux_ouverture = 0.22
taux_clic = 0.03
panier = 18

ouvertures = envois * taux_ouverture
clics = ouvertures * taux_clic
revenu = clics * panier
benefice = revenu - cout

print(f"Envois      : {envois}")
print(f"Ouvertures  : {ouvertures:.0f}")
print(f"Clics       : {clics:.0f}")
print(f"Revenu      : {revenu:.2f} euros")
print(f"Benefice    : {benefice:.2f} euros ({benefice / cout:.0%} de ROI)")

# Un rapport de gestion n'affiche jamais 2376.0000001. Le format se
# choisit selon la grandeur : des euros a deux decimales, un effectif
# sans decimale, un taux en pourcentage.
```

### Question 8 — Comparer trois campagnes

> **Votre mission :**
> - Trois campagnes ont coûté 1 500, 2 300 et 900 € et rapporté 3 000, 2 100 et 2 700 €.
> - Ranger coûts et revenus dans deux listes, puis afficher pour chacune son bénéfice et son ROI.
> - Laquelle a le meilleur ROI ? Répondez en commentaire.

```python
couts = [1500, 2300, 900]
revenus = [3000, 2100, 2700]

for i in range(3):
    benefice = revenus[i] - couts[i]
    print(f"Campagne {i + 1} : {benefice} euros de benefice, "
          f"ROI {benefice / couts[i]:.0%}")

# La troisieme : 900 euros investis, 1 800 de benefice, soit 200 % de ROI.
# La deuxieme est la seule a perdre de l'argent — et c'est pourtant celle
# qui a le plus gros budget.
```

### Question 10 — À partir de quel coût la campagne perd-elle de l'argent ?

> **Votre mission :**
> - Le revenu attendu est de **12 000 €**, quoi qu'il arrive. Seul le coût est encore négociable.
> - Calculer le ROI pour tous les coûts de **2 000 à 20 000 € par pas de 2 000**, avec une boucle `for`.
> - Tracer la courbe du ROI en fonction du coût, et y ajouter la ligne du **ROI nul** : `plt.axhline(0, color="red")`.
> - Répondez en commentaire : **à partir de quel coût le ROI devient-il négatif**, et pourquoi ce point-là ?
> - *Nouveau :* `list(range(2000, 20001, 2000))` fabrique la liste des coûts, `plt.plot(x, y)` trace la courbe.

```python
revenu = 12000
couts = list(range(2000, 20001, 2000))
rois = []

for cout in couts:
    rois.append((revenu - cout) / cout)

plt.figure(figsize=(7, 4))
plt.plot(couts, rois, marker="o")
plt.axhline(0, color="red")   # le seuil : au-dessous, on perd de l'argent
plt.title("ROI selon le cout, a revenu constant (12 000 euros)")
plt.xlabel("cout de la campagne (euros)")
plt.ylabel("ROI")
plt.show()

# Le ROI s'annule EXACTEMENT a 12 000 euros, et devient negatif des 14 000.
# Ce n'est pas un hasard : le ROI vaut (revenu - cout) / cout, donc il change
# de signe quand le cout depasse le revenu. Le graphique montre en plus ce
# qu'un tableau cache — la courbe s'effondre a gauche : les 6 000 premiers
# euros font presque tout le ROI, les suivants n'achetent plus grand-chose.
```
